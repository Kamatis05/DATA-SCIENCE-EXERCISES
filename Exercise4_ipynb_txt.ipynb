2. Country-Level Trends `(5 pts)`

Select three countries and visualize their confirmed case counts over time on the same plot.

```python
countries = ["Philippines", "US", "Brazil"]

country_trend = df[df["Country/Region"].isin(countries)]
country_trend["Date"] = pd.to_datetime(country_trend["Date"])

country_grouped = country_trend.groupby(["Date", "Country/Region"])["Confirmed"].sum().reset_index()

plt.figure()
for c in countries:
    data = country_grouped[country_grouped["Country/Region"] == c]
    plt.plot(data["Date"], data["Confirmed"], label=c)

plt.xlabel("Date")
plt.ylabel("Confirmed Cases")
plt.title("Confirmed COVID-19 Cases Over Time (Selected Countries)")
plt.legend()
plt.show()
```

3. Active vs Recovered `(5 pts)`

For a selected country, create a line chart showing Active and Recovered cases over time.

```python
# put your answer here

country = df[df["Country/Region"] == "Philippines"]
country["Date"] = pd.to_datetime(country["Date"])

plt.figure()
plt.plot(country["Date"], country["Active"], label="Active")
plt.plot(country["Date"], country["Recovered"], label="Recovered")

plt.xlabel("Date")
plt.ylabel("Cases")
plt.title("Active vs Recovered COVID-19 Cases in the Philippines")
plt.legend()
plt.show()
```

## B: Comparative Visualizations

4. Country Comparison `(5 pts)`

Using data from a single date, create a bar chart showing the top 10 countries by confirmed cases.

```python
# put your answer here

latest_date = df["Date"].max()

latest_data = df[df["Date"] == latest_date]

top10 = (
    latest_data.groupby("Country/Region")["Confirmed"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

fig, ax = plt.subplots(figsize=(12, 5))
ax.bar(top10.index, top10.values)
ax.set_title("Top 10 Countries by Confirmed Cases")
ax.set_xlabel("Country")
ax.set_ylabel("Confirmed Cases")
plt.xticks(rotation=45)
plt.show()
```

5. WHO Region Comparison `(5 pts)`

Aggregate confirmed cases by WHO Region and visualize the result using a bar chart.

```python
# put your answer here

# Aggregate confirmed cases by WHO Region
who_region_cases = df.groupby("WHO Region")["Confirmed"].sum().sort_values(ascending=False)

# Create bar chart
plt.figure()
plt.bar(who_region_cases.index, who_region_cases.values)

plt.xlabel("WHO Region")
plt.ylabel("Total Confirmed Cases")
plt.title("Total Confirmed COVID-19 Cases by WHO Region")

plt.xticks(rotation=45)
plt.show()
```


## C. Geographic Visualization

6. Geographic Spread `(10 pts)`

Using Latitude and Longitude, create a map-based visualization showing confirmed cases for a selected date.

```python
# put your answer here

import plotly.express as px
import pandas as pd

df["Date"] = pd.to_datetime(df["Date"])

selected_date = df["Date"].max()

map_data = (
    df[df["Date"] == selected_date]
    .groupby("Country/Region", as_index=False)["Confirmed"]
    .sum()
)

fig = px.choropleth(
    map_data,
    locations="Country/Region",
    locationmode="country names",
    color="Confirmed",
    hover_name="Country/Region",
    projection="natural earth",
    title=f"Global COVID-19 Confirmed Cases ({selected_date.date()})",
)

fig.show()
```

7. Regional Clustering `(15 pts)`

Create a visualization that shows how confirmed cases are distributed geographically within a single WHO Region.

```python
import plotly.express as px
import pandas as pd

region = "Western Pacific"

region_data = (
    df[df["WHO Region"] == region]
)

latest_date = region_data["Date"].max()

region_latest = (
    region_data[region_data["Date"] == latest_date]
    .groupby("Country/Region", as_index=False)["Confirmed"]
    .sum()
)

fig = px.choropleth(
    region_latest,
    locations="Country/Region",
    locationmode="country names",
    color="Confirmed",
    hover_name="Country/Region",
    projection="natural earth",
    title=f"COVID-19 Confirmed Cases in {region} Region ({latest_date.date()})",
)

fig.show()
```
